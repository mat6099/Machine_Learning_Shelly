## **Matplotlib**
It is a python library used to create visualisations. The basic process is to take data, create plot , customise plot according to our needs and then display it.

```python
from matplotlib import pyplot as plt 
# we import pyplot module which helps in plotting the desired data into a graph.
```

```python
years = [1950,1960,1970,1980,1990,2000,2010]
gdp = [300.2,543.3, 1075.9,2862.5, 5979.6,10289.7, 14958.3]


from matplotlib import pyplot as plt
plt.plot(years, gdp, color = "green", marker = "o") # to plot the line graph, marker = points style which is dots.
plt.title("Nominal GDP")
plt.xlabel("Years") #xaxis label
plt.ylabel("Billions of $") #yaxis label
plt.show()
```

## **Bar Charts**
A bar chart is useful when we want to show how a quantity varies among a discrete set of items. plt.bar() is used to create a bar chart.


```python
import matplotlib.pyplot as plt

movies = ["Annie Hall", "Ben-Hur", "Casablanca","Gandhi","West side story"]
oscars = [5,11,3,8,10]


plt.bar(movies, oscars, color = "red")
plt.title("Favorite movies")
plt.ylabel("# of Academy Awards")
plt.xlabel("Movies")
plt.show()
```

#### **Example:** Create a bar chart with subjects as x axis and marks as y axis.

```python
import matplotlib.pyplot as plt

subjects = ["Maths", "English", "Physics", "Chemistry", "Hindi"]
marks = [85,90,94,50,63]

plt.bar(subjects, marks, color = "green")
plt.title("Report Card")
plt.xlabel("Subject")
plt.ylabel("Marks")
plt.show()
```

## **Histogram**
A bar chart can also be used to visualise the distribution of numeric values by grouping the values into buckets.


#### **Example:**
 Lets take exam grades as an example. Grades are as follows: [83, 95, 91, 87, 70, 0, 85, 82, 100, 67, 73, 77, 0]. The grades are grouped into deciles: 0,10,20,..., 100 i.e. 83 goes to 80 , 67 to 60 and 95 to 90. The bar graph thus plotted becomes a histogram.


```python
import matplotlib.pyplot as plt 
from collections import Counter # to count how many times each value occurs
grades = [83, 95, 91, 87, 70, 0, 85, 82, 100, 67, 73, 77, 0]
decile = lambda grade: grade // 10*10 # function that converts each grade into its decile
# lambda is a way to write small function in a single line without usinf def.
histogram = Counter(decile(grade) for grade in grades)
# counter counts how many grades fall into each decile, for instance count of 0 is 2.

plt.bar([x for x in histogram.keys()], histogram.values(),8)
# histogram.values() gives no of students in each bucket, 8 is the width of each bar, hostogram.keys() gives deciles80,90,70 etc.
#x - 4 shifts each bar to left by 4 units, used in older veersions.
plt.axis([-5,105,0,5]) # xaxis is -5 to 105 and y axis is 0 to 5.
plt.xticks([10*i for i in range(11)]) #xaxis labels at 0,10,...,100.
plt.xlabel("Decile")
plt.ylabel(" # of students")
plt.title("Distribution of Exam 1 Grades")
plt.show()
```

### **Example**
A list of age is taken and plotted in the histogram to see the distribution of people along different ages.

```python
from collections import Counter
from matplotlib import  pyplot as pyplot

age = [20,23,24,23,25,21,26,23,22,21,24,25,20,18]
age_count = Counter(age)

plt.bar(age_count.keys(), age_count.values(), color = "pink") #keys give the key of the dictionary i.e. ages, while values give the count i.e. no of people with same age. 


plt.xlabel("Age")
plt.ylabel("Number of Students")
plt.title("Age Distribution")
plt.show()
```

### **Line Charts**

Line charts are a good choice for showing trends. plt.plot() is used to create line charts. We can make multiple calls to plt.plot() to show multiple data series on the same chart.

```python
variance = [1,2,4,8,16,32,64,128,256]
bias_squared = [256,128,64,32,16,8,4,2,1]
total_error = [x+y for x, y in zip(variance, bias_squared)]
xs = [i for i, _ in enumerate(variance)]
# i for i means for each i add i itself to the list

plt.plot(xs, variance, 'g-', label = "variance") #green solid line
plt.plot(xs, bias_squared, 'r-', label = "bias^2") # red dot dashed line
plt.plot(xs, total_error, 'b:', label = "total_error") # blue dotted line
plt.legend(loc = "best")
plt.xlabel("model complexity")
plt.title("The Bias- Variance Tradeoff")
```

### **Scatterplots**

A scatterplot is used to visualize the relationship between
two paired sets of data.

Each observation is represented by a point.

Example:

- Number of friends
- Number of minutes spent on the site per day

x-axis → number of friends

y-axis → daily minutes spent on the site


```python
friends = [70,65,72,63,71,64,60,64,67]
minutes = [175,170,205,120,220,130,105,145,190]
labels = ['a','b','c','d','e','f','g','h','i']


plt.scatter(friends, minutes)
for label, friend_count, minute_count in zip(labels, friends, minutes):
    # first iteration is ('a',70,175) and so on
    plt.annotate(label, xy = (friend_count, minute_count), xytext = (5, -5), textcoords = 'offset points')
    # xy is coordinates of that point, xytext is how much distance to keep from that coordinate i.e. 5 points horizontally and 5 points vertically.

plt.title("Daily Minutes vs. Number of friends")
plt.xlabel(" # of friends")
plt.ylabel("daily minutes spent on the site")
plt.show()
```

```python
test_1_grades = [99, 90, 85, 97, 80]

test_2_grades = [100, 85, 60, 90, 70]

plt.scatter(test_1_grades, test_2_grades)

plt.title("Axes Aren't Comparable")
plt.xlabel("test 1 grade")
plt.ylabel("test 2 grade")
plt.axis("equal") #makes the axis scale equal.


plt.show()
```
