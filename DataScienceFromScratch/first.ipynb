## Arithmetic

\+  → addition <br>
\-  → subtraction <br>
\*  → multiplication <br>
/  → division <br>
// → floor division <br>
%  → remainder <br>
** → exponentiation

```python
a = 9
b = 2

print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a//b)
print(a%b)
print(a**b)
```

```python
#use of % operator

x = 7
print('Remainder of x%2:',x%2)

if x%2 == 0:
    print('Even')
else:
    print('Odd')
```

# Function
A Function is a rule that takes inputs and produces an output.
Syntax:

def function_name(parameters): <br>
   $\hspace{1.5cm}$ statements  <br>
   $\hspace{1.5cm}$ return output <br>

Parameters - placeholder inside the function <br>
Argument - actual value passed to the function

```python
def double(x):
    return x*2

print(double(5))
```

```python
def add(a,b):
    return a+b

print(add(3,7))
```

```python
def greet(name = 'student'):
    return 'hello' + name

print(greet())
print(greet('ABC'))
```

```python
def subtract(a = 0, b= 0):
    return a - b

print(subtract(10,5))
print(subtract(b=5))
```

# Return vs Print
print() → displays something <br>
return → sends a value back from the function <br>

Returned values can be stored and used in other calculations.

```python
def double(x):
    return x * 2

result = double(5)

print(result)
print(result + 10)
```

# Functions as First Class Objects
Python allows functions to be assigned to variables and passed as arguments to other functions. <br>
Example: <br>
f = double <br>
Now f refers to the same function as double here.



```python
def double(x):
    return 2*x

f = double
print(f(5))
```

```python
def apply_to_one(f):
    return f(1)

def double(x):
    return x * 2

print(apply_to_one(double))
```

# Strings
Strings represent text. Strings can use :<br>
- single quotes <br>
- double quotes( quotes have to match) <br>

Usefule opertaions:<br>
- lower(): lowercase
- upper(): uppercase
- split(): splits the name into parts

```python
name = "Machine Learning"

print(len(name))
print(name.lower())
print(name.upper())
print(name.split())
```
