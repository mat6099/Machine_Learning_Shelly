### Variable and basic Data Types
A Variable is a name that refers to a value.
Common Python data types:
- int $\implies$ integers
- float $\implies$ decimal numbers
- str $\implies$ strings
- bool $\implies$ True/False
- None $\implies$ absence of a value

```python
x = 10
y = 3.5
name = 'Python'
is_ml = True
nothing = None

print(type(x))
print(type(y))
print(type(name))
print(type(is_ml))
print(type(nothing))
```

## Arithmetic

\+  → addition <br>
\-  → subtraction <br>
\*  → multiplication <br>
/  → division <br>
// → floor division <br>
%  → remainder <br>
** → exponentiation

```python
a = 9
b = 2

print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a//b)
print(a%b)
print(a**b)
```

```python
#use of % operator

x = 7
print('Remainder of x%2:',x%2)

if x%2 == 0:
    print('Even')
else:
    print('Odd')
```

## Modules
A module is a collection of Python functionality.
- Import an entire module: <br>
import math
- Access its functions using: <br>
math.sqrt()
- Import specific functions: <br>
from math import sqrt
- then use: <br>
sqrt()

```python
import math
print(math.sqrt(16))
print(math.pi)
```

```python
from math import sqrt, pi
print(sqrt(25))
print(pi)
```

```python
import math as m
print(m.sqrt(25))
```
