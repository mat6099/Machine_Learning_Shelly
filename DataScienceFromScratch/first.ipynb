## Control Flow

Control flow determines which statements Python executes.

Main tools:

- if : runs a block only when its condition is true.
- elif : checks another condition if earlier if condition were false.
- else : runs a block when none of the preceeding conditions were true.
- for : loops over each item in a sequence.
- while : repeats a block as long as its condition stays true.
- break : exits the loop immediately, skipping any remaining iterations.
- continue : skips the rest of the current loop iteration and jumps to the next one.

```python
# use of if-elif-else statement.
x = 10
if x > 10:
    print("x is greater than 10")
elif x ==10:  
    print("x is equal to 10")
else:
    print("x is less than 10")

```

```python
#for loop
for x in range (10):
    print(x, "is less than 10")
```

```python
#while loop
x = 0
while x < 10:
    print(x)
    x+=1 # adds 1 to each iteration
```

# Difference between for and while loop
for loop is used when number of iterations are known, while loop is used when the no of iterations are unknown. Sometimes while loop results in an infinite loop if condition never becomes false. For instance , if we do not write x+=1 in above code, python considers value of x = 0 forever and an infinite loop will start where python will  continue printing 0.

```python
# break and continue statements
for x in range(10):
    if x ==5:
        continue # skip the rest of the loop when x is 5
    

    elif x ==8:
        break # exit the loop when x is 8
    print(x)
```

### Sorting
- sorted() : returns a new sorted list
- list.sort() : sorts the existing list in place
- reverse = True : descending order
- key= : specify what should be used for sorting

```python
numbers = [4,1,3,2,5]

print(sorted(numbers)) #sorts the list in ascending order
print(numbers)
print(sorted(numbers, reverse=True)) #sorts the list in descending order

numbers.sort() #this sorts the list in place and modifies the original list
print(numbers)
# sorted function does not modify the original list, while the sort() method modifies the original list in place. sorted() accepts any iterable, while sort() is a method of lists.
```

```python
numbers = [-4,-3,-2,-1,0,1]
print(sorted(numbers, key = abs)) #sorts the list in ascending order based on absolute values
print(sorted(numbers, key = abs, reverse = True)) #sorts the list in descending order based on absolute values
```

### List Comprehensions
A compact Pythonic way to create lists.

Basic structure: <br>
[expression for item in collection] <br>
with a condition: <br>
[expression for item in collection if condition]


```python
numbers = [1,2,3,4,5]
even_numbers = [ x for x in numbers if x% 2 ==0]
even_squares = [ x**2 for x in numbers if x% 2 ==0]
print( "x%2 == 0 is true for x = ",even_numbers ,",x**2 = ",even_squares)
```

### Generator and Iterators
- A generator provides values one at a time rather than creating the entire collection at once.
- Use yield instead of return.
- Generators are useful when working with large amounts of data.
- return returns a result and exists the function while yield produces a value and pauses the function.



```python
def generate_numbers(n):
    for i in range(n):
        yield i #yield returns a generator object that can be iterated over to get the values one at a time.
for j in generate_numbers(6):
    print(j)
```

### Randomness
Random module provides functions for generating random values.

```python
# randomness provides functions for generating random values.
import random
print(random.random())
print(random.randint(1,10)) # generates a random integer between 1 and 10
```

```python
names = ["Princi", "Debasis", "Alvina","Taabir"]
print(random.choice(names)) # randomly selects a name from the list
```

### Object-Oriented Programming
Python allows us to define classes that encapsulate data and the functions that operate on them. It is useful when we do not have a built in Python set, then we can create our own set class.


```python
class Student:  #defining a class named Student
    def __init__(self, name): #initializing the class with a constructor that takes a name parameter
        self.name = name

    def introduce(self):   #method that returns a string introducing the student by name
        return "Hello, I am " + self.name

student = Student("Riya")
print(student.name)
print(student.introduce())
```

### Enumerate

### \textbf{zip()}
zip()  transforms multiple lists into a single list of tuples of corresponding elements. It combines corresponding elements from multiple sequences.

If the lists are different lengths, zip stops as soon as the first list ends.

```python
names = ["Princi", "Debasis", "Alvina","Taabir"]
scores = [85, 90, 78, 92]
for i, (name, score) in enumerate(zip(names, scores)): # enumerated and zipped the two lists together.
    print(i + 1, name, score)
```

### Argument Unpacking
Used to unzip a list. The asterisk performs argument unpacking, which uses the elements of pairs as individual arguments to zip.

```python
def add(a, b):
    return a+b
numbers = [2,3]

print(add(*numbers))  # * is used to unzip numbers list into 2 int values.



x = add(2,3) #both gives similar output , but use is different.
print(x)
```

### *args() and **kwargs()
- *args : collects unnamed positional argumnets into a tuple
- **kwargs : collects named arguments into a dictionary.

```python
def example(*args, **kwargs):
    print(args)
    print(kwargs)


example(1,2, name = "Riya",3, age = 25, gender = "Female")

#doesnt work because positional argument (3) must come before keyword argument.
```

```python
def add(*args):
    print(args) #collected arguments into a tuple
    return sum(args)

print(add(1,2,3))
print(add(5,10))
```

```python
def info(**kwargs):
    print(kwargs) #collected argumments into a dictionary.
    for key, value in kwargs.items():
        print(key, "=", value)

info(name = "Riya", age = 25, gender = "Female")
```
