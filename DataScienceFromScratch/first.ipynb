# Function
A Function is a rule that takes inputs and produces an output.
Syntax:

def function_name(parameters): <br>
   $\hspace{1.5cm}$ statements  <br>
   $\hspace{1.5cm}$ return output <br>

Parameters - placeholder inside the function <br>
Argument - actual value passed to the function

```python
def double(x):
    return x*2

print(double(5))
```

```python
def add(a,b):
    return a+b

print(add(3,7))
```

```python
def greet(name = 'student'):
    return 'hello' + name

print(greet())
print(greet('ABC'))
```

```python
def subtract(a = 0, b= 0):
    return a - b

print(subtract(10,5))
print(subtract(b=5))
```

# Return vs Print
print() → displays something <br>
return → sends a value back from the function <br>

Returned values can be stored and used in other calculations.

```python
def double(x):
    return x * 2

result = double(5)

print(result)
print(result + 10)
```

# Functions as First Class Objects
Python allows functions to be assigned to variables and passed as arguments to other functions. <br>
Example: <br>
f = double <br>
Now f refers to the same function as double here.



```python
def double(x):
    return 2*x

f = double
print(f(5))
```

```python
def apply_to_one(f):
    return f(1)

def double(x):
    return x * 2

print(apply_to_one(double))
```

# Lists:
A list is an ordered collection of values. Lists are:
- ordered
- mutable
- indexed

Indexing starts at 0.

```python
numbers = [10,20,30,40,50]
print(numbers[0])
print(numbers[-1])


numbers[0] = 70
print(numbers)


numbers.append(60) #to add an element at the end of the list
print(numbers)

numbers.remove(20) #to remove element 20 from the list
print(numbers)

print(len(numbers)) #to get the length of the list

sum_numbers = sum(numbers) # to get the sum of all elements in the list
print(sum_numbers)
```

# Tuples
A Tuple is an ordered collection similar to a list.
The difference between a list and a tuple is :
- lists are mutable while a tuple is not. 
- Lists use square brackets while tuples uses parentheses. <br>
Tuples are commonly used for fixed groups of values.

```python
tuple = (3,5)
print(tuple)

print(tuple[0])
print(tuple[-1])

x,y = tuple
print(x)
print(y)

tuple.append(7) # This will raise an error because tuples are immutable.
```

## Dictionaries

A dictionary stores key-value pairs.

Structure:

{
    key: value
}

Access values using their keys.

```python
student = {
    "name": "Riya",
    "age": 23,
    "subjects": ["Mathematics", "Physics", "Chemistry"]
}

print(student["name"])
print(student["subjects"])
print(student)
```

```python
print(student.keys())
print(student.values())
print(student.items())
```

```python
document = ["data","science", "data", "python", "latex"]

word_count = {} #empty dictionary to store word counts.
for word in document:
    if word in word_count:
        word_count[word] +=1 #if the word is already present in the dictionary, increment its count by 1
    else:
        word_count[word] = 1 #if the word is not present in the dictionary, add it with a count of 1

print(word_count)
```

# Sets
A Set is an unordered collection of unique elements. Enclosed in curly brackets.

Useful for:
- removing duplicate values.
- checking the availability of value in that set.

```python
numbers = {1,2,1,3,7,6,5,4}
print(numbers) #remove duplicates and print unique values
print(2 in numbers) #check if 2 is present in the set
print(10 in numbers)
```

## Control Flow

Control flow determines which statements Python executes.

Main tools:

- if : runs a block only when its condition is true.
- elif : checks another condition if earlier if condition were false.
- else : runs a block when none of the preceeding conditions were true.
- for : loops over each item in a sequence.
- while : repeats a block as 
- break
- continue

```python
x = 10
if x > 10:
    print("x is greater than 10")
elif x ==10:  
    print("x is equal to 10")
else:
    print("x is less than 10")

```
