## **Vectors**
A Vector is an ordered collection of numbers. In data science, we use them to represent numerical data. 
For example, the weight, height and age of a person can be represented as a three-dimensional vector:
$$
v = [70, 170, 40]
$$
where: <br>
70 - weight <br>
170 - height <br>
40 - age of a person

Python lists do not automatically perform mathematival vector operations, so we implement these operations ourselves.


### **Vector Dimension**
The dimension of a vector is the number of components it contains.
- ([3, 5]) is a two-dimensional vector.<br>
- ([95, 80, 75, 62]) is a four- dimensional vector.

```python
grades = [95,80,62,73,89]
print("Grades: ", grades)
```

### **Vector Addition**
Two vector of the same length can be added by adding their corresponding components. <br>
Mathematically, 
$$
v + w = [v_1 + w_1, ..., v_n + w_n]
$$
For example: 
$$
[1, 2] + [2,1] = [3,3]
$$
Vector of different lengths cannot be added using this component-wise operation.

```python
def vector_add(v, w):
    # Add corresponding elements of two vectors.
    return [v_i + w_i
        for v_i, w_i in zip(v, w)
    ]

print(vector_add([1, 2], [2, 1])) # [1+2, 2+1] = [3,3]
```

### **Vector Sum**
It means summing over more than two vectors componentwise.
 `vector_sum()` using the previously defined `vector_add()` function.

The approach is:

1. Start with the first vector.
2. Take each remaining vector one at a time.
3. Add it to the running result using `vector_add()`.
4. Return the final vector.

This is an example of **reusing a function**: instead of writing the componentwise addition logic again, we use the function we already created.

The vectors must all have the same length.

```python
def vector_add(v, w):
    return [ v_i + w_i
            for v_i, w_i in zip(v,w)]

def vector_sum(vectors):
    result = vectors[0] #start with first vector
    for vectors in vectors[1:]: #visit the remaining vectors
        result = vector_add(result, vectors) 

    return result
vectors = [[1,2],[3,4],[-1,1]]
print(vector_sum(vectors))
```

```python
#example correct marks in each sections A,B,C
student_1 = [4,5,6]
student_2 = [3,2,6]
student_3 = [2,4,1]
students = [student_1,student_2, student_3]
combined_answers = vector_sum(students)
print("Combined correct answers in each section is:", combined_answers)
```

### **Vector Subtraction**
Vector subtraction also works component-wise:
$$
v - w = [v_1 - w_1, ..., v_n - w_n]
$$
For example:
$$
[5,7] - [2,3] = [3,4]
$$


```python
def vector_subtract(v,w):
    #subtract corresponding elements from two vectors
    return [ v_i - w_i 
            for v_i,w_i in zip(v,w)]
print(vector_subtract([5,7], [2,3]))
```

### **Scalar Multiplication**
It means multiplying every component of a vector with the same number.
$$
cv = [cv_1, cv_2,...,cv_n]
$$
For example:
$$
3[2,4] = [6,12]
$$

### **Vector Mean**
The vector mean is calculated by taking the mean of each component across a collection of vectors. Suppose we have n vectors of equal length, their mean is
$$
\operatorname{{mean}}(v_1,\ldots,v_n)
=\frac{1}{n}\sum_{i=1}^{n}v_i.
$$

Here, the summation adds the vectors componentwise, and multiplying by (1/n) divides every component by the number of vectors.
 

```python
def vector_add(v,w):
    return [v_i +w_i for v_i,w_i in zip(v,w)]

def vector_sum(vectors):
    result = vectors[0]
    for vector in vectors[1:]:
        result = vector_add(result, vector)
    return result

def scalar_multiply(c,v):
    return[ c*v_i for v_i in v]

def vector_mean(vectors):  # sum of vectors/n
    n = len(vectors)
    return scalar_multiply(1/n, vector_sum(vectors))

vectors = [[1,2], [3,4], [5,6]]
print("Sum of vectors is:", vector_sum(vectors))
print("n = ", len(vectors))
print("Vector mean of the following vectors is :", vector_mean(vectors))
```

```python
#example
student1 = [80,70,89]
student2 = [45,60,79]
student3 = [78,67,98]
student4 = [89,90,78]

marks = [student1, student2, student3, student4]
average_marks = vector_mean(marks)
print("Average marks of the following students are:", average_marks)
```

### **Dot Product**
Dot Product of two vectors is the sum of the products of their corresponding components.\quad
Suppose
$$
v=[v_1,v_2,\ldots,v_n]
$$

and

$$
w=[w_1,w_2,\ldots,w_n].
$$

Their dot product is

$$
v\cdot w=v_1w_1+v_2w_2+\cdots+v_nw_n.
$$

The result of a dot product is a single number (a scalar), not another vector.




```python
def dot(v, w):
    return sum(v_i * w_i for v_i, w_i in zip(v, w)) #v1*w1 +v2*w2

v = [1,2]
w = [3,4]
result = dot(v,w)
print("Dot product of v and w is :", result)
```

```python
#example
#no of items purchased
items = [2,3]

#price of each item
price = [50,25]

#total cost
total_cost = dot(items, price)
print("Total Cost is : ", total_cost)

```

### **Sum of Squares**
The sum of squares of a vector is the sum of the squares of all its components.

If

$$
v=[v_1,v_2,\ldots,v_n],
$$

then

$$
\operatorname{sum\ of\ squares}(v)
=v_1^2+v_2^2+\cdots+v_n^2.
$$


Recall that the dot product is

$$
v\cdot w=\sum_{i=1}^{n}v_iw_i.
$$

If we take the dot product of a vector with itself, each component is multiplied by itself:

$$
v\cdot v=v_1^2+v_2^2+\cdots+v_n^2.
$$

Thus,

$$
\boxed{\operatorname{sum\ of\ squares}(v)=v\cdot v}.
$$

This means we do not need to write the multiplication and addition logic again. We can reuse the dot() function.


```python
def dot(v, w):
    return sum(v_i* w_i for v_i, w_i in zip(v, w))

def sum_of_squares(v):
    return dot(v,v)

v = [1,2,3]
result = sum_of_squares(v)
print(result)
```

### **Magnitude of a Vector**
The magnitude of a vector is a non-negative number that tells us how long the vector is. For a vector 
$$
v = [v_1, v_2,\dots,v_n]
$$
its magnitude is
$$
\boxed{|v|=\sqrt{v_1^2+v_2^2+\cdots+v_n^2}}.
$$
We square every component, add the squares, and then take the square root.

```python
import math
def dot(v, w):
    return sum(v_i * w_i for v_i, w_i in zip(v,w))

def sum_of_squares(v):
    return dot(v,v)

def magnitude(v):
    return math.sqrt(sum_of_squares(v))

v = [3,4]
print("Magnitude of this vector is:", magnitude(v))
```
