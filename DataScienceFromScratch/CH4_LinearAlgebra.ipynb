## **Vectors**
A Vector is an ordered collection of numbers. In data science, we use them to represent numerical data. 
For example, the weight, height and age of a person can be represented as a three-dimensional vector:
$$
v = [70, 170, 40]
$$
where: <br>
70 - weight <br>
170 - height <br>
40 - age of a person

Python lists do not automatically perform mathematival vector operations, so we implement these operations ourselves.
