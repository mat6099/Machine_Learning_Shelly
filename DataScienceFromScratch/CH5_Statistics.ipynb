## **Describing a single set of Data**
Statistics provides methods to summarize and understand a dataset.

A dataset can be described using:
1. **Central tendencies :** Mean, median, mode, and quantiles.
2. **Dispersion :** Range, variance, std deviation, and interquartile range.
3. **Relationships :** Covariance and correlation between two variables.

## **Central Tendencies:**
### **Mean**
The sum of all observations divided by the number of observations:
$$
\bar{x} = \frac{1}{n} \sum_{i=1}^{n}x_i

```python
num_friends = [2, 3, 3, 5, 7]

def mean(x):
    return sum(x) / len(x)

mean(num_friends)
```
