## **Dependence and Independence**
Probability measures uncertainity about events. An event is an outcome from a random experiment. For example, rolling a die and getting an even number is an event.
## **Independent events**
Two events (E) and (F) are independent if knowing that one occured gives us no information about whether the other occured.
Mathematically,
$$
P(E \cap F) = P(E) P(F)
$$
Here, $E\cap F$ means both the events occur.

Let us discuss about flipping a fair coin twice.<br>
- Event (E): the first flip is heads.
- Event (F): the second flip is heads.
The flips are independent and so thus the events E and F.
$$
P(E)=\frac12,\qquad P(F)=\frac12
$$

Thus ,
$$
P(E\cap F)=\frac12\times\frac12=\frac14
$$

## **Dependent Events**
Two events are dependent if knowing that one occurred gives us information about the other. <br>
Consider these events while flipping a coin twice. <br>
- Event (E): the first flip is Heads.
- Event (F):  both flips are Tails.
If the first flip is Heads, then both flip cannot be tails. Thus , these events are dependent on each other.



```python
#two fair coin flips
#E: first flip is heads
p_E = 1 / 2 
#F: second flip is heads
p_F = 1 / 2
#E and F are independent
p_EandF = p_E * p_F
print(p_EandF)


```

## **Conditional Probability**

Conditional probability is the probability of an event occurring, given that another event has already occurred.

It is written as:

$$
P(E\mid F)=\frac{P(E\cap F)}{P(F)}
$$

where $ P(F)>0$.

We now consider a family with two children

Assume each child is equally likely to be a boy or a girl, and their genders are independent.

The possible outcomes, written as (older child, younger child), are:

$$
\{BB,BG,GB,GG\}
$$

Each outcome has probability $ 1/4 $.

**Case 1: The older child is a girl.**

Given that the older child is a girl, the possible outcomes are $ GB $ and $ GG $.

Only one of these has two girls, so

$
P(\text{both girls}\mid\text{older girl})=\frac12
$

**Case 2: At least one child is a girl.**

The possible outcomes are $ BG,GB,GG $.

Only one of these has two girls, so

$$
P(\text{both girls}\mid\text{at least one girl})=\frac13
$$

```python
# let us simulate 10,000 families to check these probabilities.
import random

def random_kid():
    return random.choice(["boy", "girl"])

both_girls = 0
older_girl = 0
either_girl = 0

random.seed(0)
for _ in range(10000):
    younger = random_kid()
    older = random_kid()

    if older == "girl":
        older_girl+=1

    if older =="girl" and younger =="girl":
        both_girls +=1

    if older == "girl" or younger =="girl":
        either_girl +=1

print("P(both | older):", both_girls / older_girl)
print("P(both | either):", both_girls / either_girl)


```

## **Bayes's Theorem**
Bayes's Theorem helps us calculate the probability of an event given new information, using a conditional probability in the reverse direction.
**Formula**
$$
P(E\mid F)=
\frac{P(F\mid E)P(E)}{P(F)}
$$
$$
P(E\mid F)=
\frac{P(F\mid E)P(E)}
{P(F\mid E)P(E)+P(F\mid E^c)P(E^c)}
$$
where $E^c$ means that event $E$ does not occur.

We take an example of a disease and a test with these properties:

$P(D)=0.0001$: 1 in 10,000 people have the disease.

$P(T\mid D)=0.99$: 99% of people with the disease test positive.

$P(T\mid D^c)=0.01$: 1% of people without the disease test positive.

$ P( D^c)=0.9999 $.

We want to find $ P(D\mid T)$: the probability that a person has the disease, given a positive test.




```python
# Probabilities from the book
p_disease = 0.0001
p_no_disease = 0.9999

p_positive_given_disease = 0.99
p_positive_given_no_disease = 0.01

# Bayes's Theorem
p_disease_given_positive = (
    p_positive_given_disease * p_disease
) / (
    p_positive_given_disease * p_disease
    + p_positive_given_no_disease * p_no_disease
)

print(p_disease_given_positive)
print(p_disease_given_positive * 100)
```

## **Expected value**
The expected value is the average of a random variable's possible values, weighted by their possibilities.
$$
E[X] = \sum_x xP(X=x)
$$
For the coin-flip variable:<br>
$$
E[X] = 0(0.5) + 1(0.5) = 0.5
$$
For the variable that takes values from 0 to 9 with equal probability:
$$
E[X] = \frac{0 + 1 + \cdots + 9}{10} = 4.5
$$


```python
values = [0, 1]
probabilities = [0.5, 0.5]

expected_value = sum(
    value* probability
    for value, probability in zip(values, probabilities)
)
print(expected_value)
```

```python
values = list(range(10))
probability = 0.1

expected_value = sum(value* probability for value in values)
print(expected_value) 
```

Let $X$ be the number of girls. Its distribution is: <br>
- $P(X=0)=1/4$
- $P(X=1)=1/2$
- $P(X=2)=1/4$

If we know at least one child is girl, the conditional variable $Y$ has probabilities $ P(Y = 1) = 2/3 $ and $ P(Y = 2) = 1/3 $.

If we know the older child is a girl, the conditional variable $Z$ has probabilities $P(Z=1)=1/2$ and $P(Z=2)=1/2$.

## **Continuous Distributions**

A **continuous distribution** models a random variable that can take values across a continuum, such as any real number between 0 and 1.

### 1. **Uniform distribution**

The book uses the **uniform distribution** on the interval $[0,1]$, where all values are equally weighted.

Its probability density function (PDF) is:

$$
f(x)=
\begin{cases}
1, & 0\le x<1\\
0, & \text{otherwise}
\end{cases}
$$

### 2. **Probability density function (PDF)**

For a continuous random variable, the probability of an individual exact value is zero. Instead, probabilities are assigned to intervals.

The probability that $X$ lies in an interval is the area under the PDF over that interval.

For the uniform distribution:

$$
P(0.2\le X\le 0.3)=0.3-0.2=0.1
$$

### 3. **Cumulative distribution function (CDF)**

The **CDF** gives the probability that a random variable is less than or equal to a specified value:

$$
F(x)=P(X\le x)
$$

For the uniform distribution on $[0,1]$:

$$
F(x)=
\begin{cases}
0, & x<0\\
x, & 0\le x<1\\
1, & x\ge1
\end{cases}
$$



```python
def uniform_pdf(x):
    return 1 if x >= 0 and x < 1 else 0

print(uniform_pdf(0.4))   
print(uniform_pdf(1.2))  
```

```python
def uniform_cdf(x):
    #  Probability that a uniform random variabl is <=x.
    if x < 0:
        return 0
    elif x < 1:
        return x
    else:
        return 1

print(uniform_cdf(-0.2))
print(uniform_cdf(0.4))
print(uniform_cdf(1.5))
```

## **The Normal Distribution**
The normal distribution is a bell-shaped probability distribution determined by two parameters:

- Mean $\mu $: determines the centre of the curve.

- Standard deviation $ \sigma $ : determines how wide or narrow the curve is.

### **1. Probability density function (PDF)**
The normal distribution's PDF is

$$
f(x;\mu,\sigma)=
\frac{1}{\sqrt{2\pi}\sigma}
e^{-\frac{(x-\mu)^2}{2\sigma^2}}
$$

Here, $x$ is the value whose density we want to calculate, $\mu $ is the mean, and $\sigma$ is the standard deviation.

### **2. Standard normal distribution**
The standard normal distribution has mean (0) and standard deviation (1).

Any normal distribution can be standardized using

$$
z=\frac{x-\mu}{\sigma}
$$

The (z)-score tells us how many standard deviations a value lies above or below the mean.

### **3. Cumulative Distribution Function (CDF)**
The CDF gives the probability that a normally distributed random variable is less than or equal to (x):

$$
F(x)=P(X\leq x)
$$

Unlike the PDF, the CDF directly gives cumulative probabilities.





```python
import math

def normal_pdf(x, mu=0, sigma=1):
    sqrt_two_pi = math.sqrt(2 * math.pi)
    return math.exp(-(x - mu) ** 2 / 2 / sigma ** 2) / (
        sqrt_two_pi * sigma
    )

print(normal_pdf(0))  # approximately 0.3989
print(normal_pdf(1))  # approximately 0.2420
```

```python
def normal_cdf(x, mu=0, sigma=1):
    return (
        1 + math.erf((x - mu) / math.sqrt(2) / sigma)
    ) / 2

print(normal_cdf(0))  # 0.5
```

```python
def normal_probability_between(lo, hi, mu = 0, sigma = 1):
    return normal_cdf(hi, mu, sigma) - normal_cdf(lo, mu, sigma)

print(normal_probability_between(-1,1))
```

```python
# plot different normal PDFs
import matplotlib.pyplot as plt

xs = [x / 10 for x in range(-50, 50)]

plt.plot(xs, [normal_pdf(x, sigma=1) for x in xs],
         label="mu=0, sigma=1")
plt.plot(xs, [normal_pdf(x, sigma=2) for x in xs],
         label="mu=0, sigma=2")
plt.plot(xs, [normal_pdf(x, sigma=0.5) for x in xs],
         label="mu=0, sigma=0.5")
plt.plot(xs, [normal_pdf(x, mu=-1) for x in xs],
         label="mu=-1, sigma=1")

plt.legend()
plt.title("Various Normal PDFs")
plt.show()
```
